# Minimum Working Example

This is an MWE to verify that the API connection works.

Import packages.

_Note that first time you may need to restart your Jupyter IDE a couple of times to make it work!_

Make sure that the path to the directory with lseg-data.config.json file is correct and the file contains the API key, username, and password.

Open the session

```python
import lseg.data as ld
import os

os.environ["LD_LIB_CONFIG_PATH"] = "../../Configuration"
ld.open_session()
```

Run your code

```python
df = ld.get_history(universe="LSEG.L")
print(df.head())
```
