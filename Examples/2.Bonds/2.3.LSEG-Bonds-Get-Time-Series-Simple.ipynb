# Bonds: Get time series - simple approach

```python
import lseg.data as ld
import os
import pandas as pd

os.environ["LD_LIB_CONFIG_PATH"] = "../../Configuration"
ld.open_session()
```

The easiest way to get the time series is to use get_history()

```python
start_date = "2026-01-01"
end_date = "2026-03-11"
ric = "US315890934="

df = ld.get_history(universe=ric,
                    interval="1D",
                    start=start_date,
                    end=end_date)
df
```
